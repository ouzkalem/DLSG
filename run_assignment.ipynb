## 2. Sanity Check - Küçük Veri Seti ile Test

```python
# Küçük veri seti hazırla
train_loader, test_loader, stats = prepare_data(
    n_samples=DATA_CONFIG["n_samples_small"],
    batch_size=TRAIN_CONFIG["batch_size"],
)

# Model oluştur
model = create_model("standard")
print(model)

# Optimizer ve Loss
optimizer = torch.optim.Adam(model.parameters(), lr=TRAIN_CONFIG["learning_rate"])
criterion = torch.nn.MSELoss()

# Eğitim
print("\n" + "="*60)
print("  SANITY CHECK: Küçük veri seti ile eğitim")
print("="*60)

history = train_model(model, train_loader, test_loader, criterion, optimizer, 
                      epochs=TRAIN_CONFIG["epochs"], verbose=True)

# Sonuç
print(f"\n✓ Sanity check başarılı!")
print(f"  Final Train Loss: {history['train_loss'][-1]:.6f}")
print(f"  Final Test Loss: {history['test_loss'][-1]:.6f}")

# Grafik
plot_loss_curves(history, title="Sanity Check")
```
